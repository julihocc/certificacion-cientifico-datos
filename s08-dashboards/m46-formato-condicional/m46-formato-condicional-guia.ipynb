Trabajas sobre el modelo estrella del Módulo 45, que [`../practica-ventas/ventas.py`](../practica-ventas/ventas.py) abre en DuckDB con `conectar()`. La vista `ventas_detalle` une la tabla de hechos (1 499 000 ventas de 2019 a 2021, con `fecha`, `cantidad`, `precio` e `importe` = `cantidad` × `precio`) con sus dimensiones (`pais`, `categoria`, `anio`...).

No hace falta escribir SQL: estas funciones de `ventas.py` ya calculan lo que necesitas, y todas aceptan el filtro `paises` (una lista; `None` significa "sin filtro" y una lista vacía no deja pasar ninguna fila):

| Función | Devuelve |
| --- | --- |
| `kpis(con, paises=...)` | Diccionario con `ventas_total`, `cantidad_total`, `precio_mediana`, `ventas_promedio` y `transacciones` |
| `ventas_por(con, "categoria", paises=...)` | Ventas y cantidad por una o varias columnas de las dimensiones |
| `tabla_por_pais(con, paises=...)` | Por país: `iso3`, `ventas_mediana`, `ventas_total`, `cantidad_total` y `pct_cantidad` (fracción de la cantidad total) |
| `serie_temporal(con, nivel, paises=...)` | Ventas por `periodo`, con `nivel` igual a `"day"`, `"month"` o `"year"` |
