## Contexto

La dirección comercial de la cadena de tiendas quiere una primera página que resuma de un vistazo las ventas de 2019 a 2021: cuánto se vendió en total, en qué países, qué categorías de producto pesan más y cómo evolucionaron las ventas en el tiempo.

Además pide una tabla por país, pero no quiere leer cuarenta números: el país que más vende, el que menos y las diferencias entre ellos deben saltar a la vista gracias al color.
