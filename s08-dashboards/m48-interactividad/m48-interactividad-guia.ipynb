## Contexto

Los gerentes de la cadena de tiendas ya usan el tablero de ventas, pero quieren hacer sus propias preguntas sin pedir un informe nuevo cada vez: acotar el análisis a un periodo concreto, ver la evolución de un país con solo pasar el cursor por el mapa y, cuando algo les llama la atención, abrir el detalle de los productos vendidos en ese país.

Tu tarea es añadir al informe una página de análisis con esas tres formas de interacción, sin recargar la página principal.

## Tarea 4. Obtención de detalles

3. Comprueba que Power BI añade un botón **Atrás** para volver a la página de origen.
4. En el mapa de la página de análisis, haz clic derecho sobre un país y elige **Obtener detalles → drill-through**. Comprueba que la tabla muestra solo los productos de ese país.

*Describe aquí los pasos y añade una captura de pantalla.*
