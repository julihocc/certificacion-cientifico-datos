## Contexto

La app de ventas del Módulo 49 está terminada y la dirección quiere compartirla con los gerentes regionales. El gerente de Norteamérica, sin embargo, solo debe ver las ventas de los países del antiguo tratado NAFTA (Estados Unidos, Canadá y México), aunque use la misma app que el resto.

Mantener una app distinta por región sería costoso y propenso a errores. La alternativa es publicar una sola app y filtrar las filas según quién la mire: con roles fijos (seguridad estática) o con una tabla que asigna países a cada correo (seguridad dinámica).

| Columna | Valores |

Todas las medidas de [`ventas.py`](../practica-ventas/ventas.py) aceptan el argumento `paises`: `None` no filtra, una lista deja solo esos países y una lista vacía no deja ninguna fila.
