En este ejercicio vas a convertir las páginas sueltas de los módulos anteriores en un informe coherente y fácil de recorrer: un tema común, una página de menú, botones para volver al inicio, marcadores y navegadores de páginas y de marcadores.

## Contexto

El informe de ventas de la cadena de tiendas ya tiene varias páginas: el dashboard principal, la validación de medidas DAX, la página de análisis con el segmentador de fechas y las páginas auxiliares de detalle e información sobre herramientas. Pero cada una tiene su propio estilo y el usuario no sabe por dónde empezar ni cómo volver.

Antes de distribuir el informe, la dirección pide que tenga una portada con un menú, un aspecto uniforme y accesos rápidos a los periodos que más se consultan, como los dos semestres de 2019.

## Tarea 4. Marcadores y navegadores

3. Agrega un **navegador de marcadores** con esos dos botones.
4. Agrega un **navegador de páginas** con forma de flecha (chevron) para moverte entre las páginas principales, y el logotipo.
