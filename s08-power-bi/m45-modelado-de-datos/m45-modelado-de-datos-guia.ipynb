## Contexto

Una cadena de tiendas con presencia en diez países de América y Europa ya tiene sus ventas de 2019 a 2021 y sus catálogos de clientes, productos, tiendas y países cargados en Power BI. Por separado, esas tablas no responden preguntas como "¿cuánto vendió cada categoría en Canadá?" o "¿qué tiendas venden más muebles?".

Antes de construir ningún informe, el equipo de análisis necesita **relacionar** las tablas para que un filtro sobre un país, un producto, un cliente o una tienda llegue a las ventas en un solo salto, sin ambigüedades.
