## Contexto

El informe de ventas está terminado y la dirección quiere compartirlo con los gerentes regionales. El gerente de Norteamérica, sin embargo, solo debe ver las ventas de los países del antiguo tratado NAFTA (Estados Unidos, Canadá y México), aunque use el mismo informe que el resto.

Mantener un informe distinto por región sería costoso y propenso a errores. La alternativa es publicar un único informe y definir roles de seguridad que filtren las filas según quién lo mire.
