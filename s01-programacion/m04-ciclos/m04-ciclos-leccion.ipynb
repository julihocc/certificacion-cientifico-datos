## 1. El ciclo `while`

`while` repite un bloque **mientras** su condición sea verdadera. Antes de cada vuelta se vuelve a evaluar la condición; en cuanto es falsa, el programa continúa después del bloque.

Dentro del bloque algo debe cambiar para que la condición llegue a ser falsa. Si nada cambia, el ciclo nunca termina: es un **bucle infinito**, y en un notebook se detiene con el botón de interrumpir el kernel.

```python
cuenta_regresiva = 5

while cuenta_regresiva > 0:
    print(cuenta_regresiva, end=" ")
    cuenta_regresiva = cuenta_regresiva - 1   # sin esta línea, el ciclo no terminaría
print("¡despegue!")
```

`while` es la opción natural cuando no se sabe cuántas vueltas habrá. Por ejemplo, ¿cuántos años tarda en duplicarse un ahorro al 7 % anual?

```python
saldo = 1000
anios = 0
while saldo < 2000:
    saldo = saldo * 1.07
    anios += 1
print(f"Se duplica en {anios} años (saldo: {saldo:.2f})")
```

## 2. Contadores y acumuladores

Dos patrones aparecen en casi todos los ciclos:

| Patrón | Valor inicial | Actualización | Uso |
| --- | --- | --- | --- |
| Contador | `0` | `contador += 1` | Cuántas veces ocurre algo |
| Acumulador de suma | `0` | `total += x` | Suma de valores |
| Acumulador de producto | `1` | `producto *= x` | Producto de valores |

El valor inicial es el **elemento neutro** de la operación: sumar 0 o multiplicar por 1 no cambia el resultado. Iniciar un producto en 0 lo deja siempre en 0.

```python
ventas = [120, 0, 340, 85, 0, 410]

total = 0          # acumulador
dias_sin_venta = 0 # contador
for v in ventas:
    total += v
    if v == 0:
        dias_sin_venta += 1

print("Total vendido:", total)
print("Días sin venta:", dias_sin_venta)
```

## 3. El ciclo `for` con `range`

`for` recorre los elementos de una secuencia, uno por vuelta. Cuando se necesita una serie de enteros se usa `range`:

| Llamada | Valores generados |
| --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(10, 0, -3)` | 10, 7, 4, 1 |

Igual que en el slicing, el límite final **no se incluye**. Por eso, para sumar de 1 a $n$ se escribe `range(1, n + 1)`.

```python
n = 10
suma = 0
for i in range(1, n + 1):
    suma += i
print("Suma de 1 a 10:", suma)
print("Fórmula n(n+1)/2:", n * (n + 1) // 2)

print(list(range(10, 0, -3)))
```

Regla práctica: `for` cuando se sabe cuántas vueltas habrá o se recorre una colección; `while` cuando el ciclo depende de una condición que cambia sobre la marcha.

## 4. `break`, `continue` y `return` temprano

- `break` sale del ciclo de inmediato.
- `continue` salta el resto de la vuelta actual y pasa a la siguiente.
- Dentro de una función, `return` termina la función entera, y con ella el ciclo. Es la forma más limpia de responder "sí" o "no" en cuanto se conoce la respuesta.

```python
# break: detenerse en el primer múltiplo de 7 mayor que 50
for k in range(51, 100):
    if k % 7 == 0:
        print("Primer múltiplo de 7 después de 50:", k)
        break

# continue: sumar solo los números impares
suma_impares = 0
for k in range(1, 11):
    if k % 2 == 0:
        continue
    suma_impares += k
print("Suma de impares de 1 a 10:", suma_impares)
```

```python
def contiene_negativo(valores):
    for v in valores:
        if v < 0:
            return True    # en cuanto aparece uno, no hace falta seguir
    return False           # se llega aquí solo si el ciclo terminó sin encontrar

print(contiene_negativo([3, 8, -1, 5]))
print(contiene_negativo([3, 8, 1, 5]))
```

## 5. Validación de entradas

Un ciclo que espera un entero positivo puede comportarse mal con otra cosa: con un decimal puede dar un resultado sin sentido o no terminar nunca, y con un texto lanza un error. Conviene validar **antes** de entrar al ciclo:

- `isinstance(x, int)` comprueba el tipo. Ojo: `True` y `False` también son `int` en Python.
- Para aceptar floats con valor entero, como `6.0`, se puede usar `float(x).is_integer()`.

```python
def es_entero_no_negativo(x):
    if isinstance(x, bool):                 # excluir True/False
        return False
    if isinstance(x, int):
        return x >= 0
    if isinstance(x, float):
        return x.is_integer() and x >= 0    # acepta 6.0, rechaza 6.5
    return False

for valor in [6, 6.0, 6.5, -2, "6", True]:
    print(repr(valor), "->", es_entero_no_negativo(valor))
```

## 6. Traza a mano de un ciclo

Antes de confiar en un ciclo conviene **trazarlo**: anotar el valor de cada variable en cada vuelta. Tomemos un ciclo que cuenta cuántas veces se puede dividir un número entre 2 hasta llegar a 1:

| Vuelta | `n` al empezar | ¿`n > 1`? | `n` al terminar | `pasos` |
| --- | --- | --- | --- | --- |
| 1 | 8 | sí | 4 | 1 |
| 2 | 4 | sí | 2 | 2 |
| 3 | 2 | sí | 1 | 3 |
| — | 1 | no: sale | — | 3 |

Así, $8 = 2^3$: tres divisiones. Imprimir las variables dentro del ciclo es la versión automática de la misma traza.

```python
n = 8
pasos = 0
while n > 1:
    n = n // 2
    pasos += 1
    print(f"vuelta {pasos}: n = {n}")
print("Divisiones:", pasos)
```

## 7. Casos borde

Un ciclo se diseña pensando en el caso típico, pero debe probarse también en los extremos: 0, 1, valores negativos y colecciones vacías. La pregunta clave es **qué pasa si el ciclo no da ni una vuelta**. En ese caso la función devuelve lo que haya después del ciclo, y ese valor debe ser correcto también para esas entradas.

```python
def suma_hasta(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

# Con n = 0 o negativo, range(1, n + 1) está vacío y la función devuelve 0
print(suma_hasta(4), suma_hasta(0), suma_hasta(-3))
```

Aquí el 0 es razonable para $n = 0$, pero para $n = -3$ quizá habría que avisar de una entrada inválida. Decidirlo es parte del diseño.

## 8. Errores frecuentes

**Un `while n > 1` que no se ejecuta con entradas pequeñas.** Si una función decide "sí" o "no" dentro del ciclo y devuelve un valor por defecto al salir, ese valor también se devuelve cuando el ciclo no da ninguna vuelta. Con $n \le 1$ la condición es falsa desde el principio y la respuesta puede ser incorrecta. Por ejemplo, para saber si un número es potencia de 10:

```python
def es_potencia_de_10_mal(n):
    while n > 1:
        if n % 10 != 0:
            return False
        n = n // 10
    return True                    # también se llega aquí con 0 y negativos

def es_potencia_de_10(n):
    if n <= 0:                     # ninguna potencia de 10 es 0 o negativa
        return False
    while n > 1:
        if n % 10 != 0:
            return False
        n = n // 10
    return True                    # n == 1 es correcto: 10**0 == 1

for valor in [1000, 1, 250, 0, -100]:
    print(valor, es_potencia_de_10_mal(valor), es_potencia_de_10(valor))
```

**Olvidar actualizar la variable de control.** Si el cuerpo de un `while` no modifica la variable de la condición, el ciclo es infinito. Una protección útil mientras se depura es un límite de vueltas:

```python
x = 10
vueltas = 0
while x != 0 and vueltas < 100:   # límite de seguridad
    x -= 3                        # 10, 7, 4, 1, -2, ... nunca vale 0 exactamente
    vueltas += 1
print("x =", x, "tras", vueltas, "vueltas")
```

La condición `x != 0` nunca se cumple porque `x` salta de 1 a −2. Una condición como `x > 0` es más robusta.

**Iniciar el acumulador dentro del ciclo.** Si `total = 0` está dentro del `for`, se reinicia en cada vuelta y al final solo contiene el último valor. La inicialización va siempre antes del ciclo.

**Errores de uno en `range`.** `range(1, n)` llega hasta $n - 1$. Para incluir $n$ se escribe `range(1, n + 1)`.

## 9. Ejercicio

La **conjetura de Collatz** dice que, partiendo de cualquier entero positivo $n$ y aplicando repetidamente la regla

$$n \to \begin{cases} n / 2 & \text{si } n \text{ es par} \\ 3n + 1 & \text{si } n \text{ es impar} \end{cases}$$

siempre se llega a 1.

1. Escribe una función `pasos_collatz(n)` que cuente, con un `while`, cuántos pasos hacen falta para llegar a 1.
2. Si `n` no es un entero positivo (0, negativo, decimal, texto o booleano), debe devolver `"Entrada inválida"`.
3. Pruébala con 1, 6, 27, 0, -5, 4.5, "8" y `True`. ¿Cuántos pasos da `pasos_collatz(1)`? Comprueba que el resultado tiene sentido aunque el ciclo no se ejecute.
4. Con un `for`, encuentra el número entre 1 y 100 que necesita más pasos.

### Solución

```python
def pasos_collatz(n):
    if isinstance(n, bool) or not isinstance(n, int) or n < 1:
        return "Entrada inválida"
    pasos = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        pasos += 1
    return pasos

for valor in [1, 6, 27, 0, -5, 4.5, "8", True]:
    print(repr(valor), "->", pasos_collatz(valor))
```

```python
mejor_n, mas_pasos = 1, 0
for k in range(1, 101):
    p = pasos_collatz(k)
    if p > mas_pasos:
        mejor_n, mas_pasos = k, p
print(f"Entre 1 y 100, el que más pasos necesita es {mejor_n} ({mas_pasos} pasos)")
```

Con $n = 1$ el ciclo no da ninguna vuelta y la función devuelve 0 pasos, lo cual es correcto: ya se está en 1. Los valores 0 y negativos se rechazan antes del ciclo; con 0, la regla daría 0 → 0 → 0... y el `while n != 1` no terminaría nunca.

## Resumen

- `while` repite mientras su condición sea verdadera; algo dentro del ciclo debe hacerla falsa para evitar un bucle infinito.
- `for` con `range` recorre una serie de enteros; el límite final no se incluye.
- Los contadores empiezan en 0 y los acumuladores en el elemento neutro de su operación (0 para sumas, 1 para productos).
- `break`, `continue` y un `return` temprano controlan cuándo termina el ciclo o la función.
- Trazar el ciclo a mano y probar los casos borde (0, 1, negativos, entradas vacías) revela errores que el caso típico oculta.
