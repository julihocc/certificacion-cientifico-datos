## 1. Booleanos y comparaciones

Toda decisión se apoya en un valor **booleano**: `True` o `False`. Los operadores de comparación producen booleanos y los operadores lógicos los combinan.

| Operador | Significado | Ejemplo | Resultado |
| --- | --- | --- | --- |
| `==` | igual a | `3 == 3` | `True` |
| `!=` | distinto de | `3 != 4` | `True` |
| `<`, `<=` | menor, menor o igual | `5 <= 5` | `True` |
| `>`, `>=` | mayor, mayor o igual | `2 > 7` | `False` |
| `and` | ambas son verdaderas | `True and False` | `False` |
| `or` | al menos una es verdadera | `True or False` | `True` |
| `not` | niega el valor | `not True` | `False` |

````python
temperatura = 31

print(temperatura > 30)                        # comparación simple
print(temperatura > 30 and temperatura < 35)   # ambas condiciones
print(not temperatura > 30)                    # negación
print(type(temperatura > 30))                  # el resultado es un bool
````

## 2. La instrucción `if` y la indentación

`if` ejecuta un bloque solo cuando su condición es verdadera. La línea del `if` termina en dos puntos y el bloque que depende de ella va **indentado** (por convención, cuatro espacios). En Python la indentación no es estética: es la que indica dónde empieza y dónde termina el bloque.

````python
saldo = 1200
retiro = 500

if retiro <= saldo:
    saldo = saldo - retiro          # dentro del if
    print("Retiro autorizado")      # dentro del if
print(f"Saldo final: {saldo}")      # fuera del if: se ejecuta siempre
````

## 3. `elif` y `else`

Cuando hay más de dos caminos, se encadenan varias condiciones con `elif` ("si no, si...") y se cierra con `else`, que recoge todo lo que no cumplió ninguna condición anterior. Python evalúa las condiciones **de arriba abajo** y ejecuta **solo el primer bloque** cuya condición sea verdadera; el resto se ignora.

````python
def clasificar_imc(imc):
    if imc < 18.5:
        return "bajo peso"
    elif imc < 25:
        return "normal"
    elif imc < 30:
        return "sobrepeso"
    else:
        return "obesidad"

for valor in [17.2, 22.0, 27.5, 31.8]:
    print(valor, "->", clasificar_imc(valor))
````

La segunda condición es solo `imc < 25` y no `imc >= 18.5 and imc < 25`. No hace falta repetir el límite inferior: si el programa llega a ese `elif`, ya sabe que la primera condición fue falsa.

## 4. Condiciones compuestas y encadenadas

Para comprobar que un valor está dentro de un intervalo se pueden unir dos comparaciones con `and`. Python admite además la **comparación encadenada**, que se lee como en matemáticas: `a < x <= b` equivale a `a < x and x <= b`.

Por ejemplo, el intervalo $(18, 65]$ se escribe `18 < edad <= 65`.

````python
edad = 40

print(edad > 18 and edad <= 65)   # forma con and
print(18 < edad <= 65)            # forma encadenada, equivalente y más legible

# Fuera del intervalo
print(18 < 70 <= 65)
````

## 5. El orden de las condiciones

Como solo se ejecuta la primera rama verdadera, el orden importa. Si una condición general aparece antes que una más específica, la específica nunca llega a evaluarse: es una **rama inalcanzable**. La regla práctica es ir de lo más restrictivo a lo menos restrictivo, o bien recorrer los intervalos en orden.

````python
def descuento_mal_ordenado(monto):
    if monto > 100:
        return 0.05
    elif monto > 500:      # inalcanzable: todo monto > 500 ya entró en la rama anterior
        return 0.10
    else:
        return 0.0

def descuento(monto):
    if monto > 500:        # primero el caso más restrictivo
        return 0.10
    elif monto > 100:
        return 0.05
    else:
        return 0.0

print(descuento_mal_ordenado(800), descuento(800))
````

## 6. Validación de entradas

Antes de calcular conviene comprobar que los datos tienen sentido. Una cantidad negativa de productos o una edad de 300 años indican un error de captura. Lo recomendable es poner esa comprobación **al principio**, como primera rama, para que los casos válidos no tengan que preocuparse por ella.

````python
def precio_total(cantidad, precio_unitario):
    if cantidad < 0 or precio_unitario < 0:
        return "Entrada inválida"
    return cantidad * precio_unitario

print(precio_total(3, 12.5))
print(precio_total(-2, 12.5))
````

## 7. La expresión condicional

Cuando solo se elige entre dos **valores**, se puede usar la expresión condicional en una sola línea:

```python
valor_si_verdadero if condición else valor_si_falso
```

Es útil para asignaciones sencillas. Si cada rama necesita varias instrucciones, es mejor el `if` de varias líneas.

````python
calificacion = 7.4
estado = "aprobado" if calificacion >= 6 else "reprobado"
print(estado)

numero = -9
valor_absoluto = numero if numero >= 0 else -numero
print(valor_absoluto)
````

## 8. Reutilizar la lógica con una función

En los ejemplos anteriores ya aparece `def`. Envolver las condiciones en una **función** permite aplicar la misma regla a muchos valores sin copiar el código. La función recibe datos como parámetros y devuelve el resultado con `return`. Las funciones se estudian a fondo en el Módulo 05; por ahora basta con este uso.

````python
def tarifa_estacionamiento(horas):
    """Primera hora a 20, las siguientes a 15; más de 8 horas, tarifa fija de 140."""
    if horas < 0:
        return "Entrada inválida"
    elif horas <= 1:
        return 20
    elif horas <= 8:
        return 20 + (horas - 1) * 15
    else:
        return 140

for h in [0.5, 3, 8, 10, -1]:
    print(f"{h} h -> {tarifa_estacionamiento(h)}")
````

## 9. Errores frecuentes

**Confundir `=` con `==`.** `=` asigna un valor; `==` compara. Escribir `if x = 5:` produce un `SyntaxError`.

**Escribir la condición como en el lenguaje hablado.** `x == 1 or 2` no significa "x es 1 o 2": Python lo lee como `(x == 1) or 2`, y como `2` es un valor verdadero, la condición siempre se cumple.

````python
x = 7
if x == 1 or 2:
    print("Forma incorrecta: entra aunque x vale", x)

if x == 1 or x == 2:
    print("No se imprime")
else:
    print("Forma correcta: x no es 1 ni 2")

print(x in (1, 2))   # alternativa compacta
````

**Dejar huecos entre intervalos.** Si una rama usa `< 10` y la siguiente `> 10`, el valor exacto 10 no entra en ninguna y cae en el `else`, que quizá estaba pensado para otra cosa. Cada frontera debe pertenecer exactamente a un intervalo.

````python
def nivel_con_hueco(puntos):
    if puntos < 10:
        return "bajo"
    elif puntos > 10:
        return "alto"
    else:
        return "error"      # aquí cae el 10 sin querer

def nivel(puntos):
    if puntos < 10:
        return "bajo"
    else:
        return "alto"       # el 10 queda incluido en "alto"

print(nivel_con_hueco(10), nivel(10))
````

**Un `else` que nunca se alcanza.** Si las ramas anteriores ya cubren todos los valores posibles (por ejemplo, `x <= 0` y `x > 0`), el `else` final es código muerto. Si la intención es atrapar entradas inválidas, la validación debe ir primero, como en la sección 6.

## 10. Ejercicio

Una paquetería cobra el envío según el peso del paquete, en kilogramos:

| Peso | Costo |
| --- | --- |
| hasta 1 kg | 90 |
| más de 1 kg y hasta 5 kg | 90 más 25 por cada kg por encima de 1 |
| más de 5 kg | 190 más 18 por cada kg por encima de 5 |

1. Escribe una función `costo_envio(peso)` que aplique la tabla usando comparaciones encadenadas.
2. Si el peso es 0 o negativo, debe devolver el texto `"Peso inválido"`.
3. Añade un parámetro `express` (booleano): si es verdadero, el costo aumenta un 30 %. Usa una expresión condicional.
4. Prueba la función con 0.8, 1, 3, 5, 7.5 y -2 kg, con y sin envío express.

### Solución

````python
def costo_envio(peso, express=False):
    if peso <= 0:
        return "Peso inválido"
    elif peso <= 1:
        costo = 90
    elif 1 < peso <= 5:     # el límite inferior es redundante, pero documenta el intervalo
        costo = 90 + (peso - 1) * 25
    else:
        costo = 190 + (peso - 5) * 18

    factor = 1.3 if express else 1.0
    return round(costo * factor, 2)

for p in [0.8, 1, 3, 5, 7.5, -2]:
    print(f"{p:>4} kg  normal: {costo_envio(p)}  express: {costo_envio(p, express=True)}")
````

## Resumen

- Las comparaciones producen booleanos, y `and`, `or` y `not` los combinan.
- `if`, `elif` y `else` eligen un único bloque: el primero cuya condición es verdadera, evaluando de arriba abajo.
- Las comparaciones encadenadas (`a < x <= b`) expresan intervalos con claridad; cada frontera debe pertenecer a un solo intervalo.
- El orden de las ramas importa: una condición general colocada antes deja inalcanzables las más específicas.
- La validación de entradas va al principio, y la expresión `a if c else b` resuelve elecciones simples entre dos valores.
