## Contexto

Una empresa paga por hora y quiere calcular el pago semanal con horas extra escalonadas: hasta 40 h a tarifa normal, de la hora 41 a la 50 a la tarifa × 1.1 y a partir de la hora 51 a la tarifa × 1.2. Por ejemplo, 55 h con tarifa 5 dan $40(5) + 10(5)(1.1) + 5(5)(1.2) = 285$.

## Instrucciones

1. **Función de cálculo del pago.** Escribir `calculo(horas, tarifa)` con `if`/`elif`/`else` según los tres tramos.
2. **Comprobar con el ejemplo** de 55 horas y tarifa 5.
3. **Casos de prueba:** 35 h con tarifa 10, 47 h con tarifa 8 y 65 h con tarifa 15.

*El enunciado original del curso no se conserva; estas instrucciones se reconstruyeron a partir de la solución.*

### Tarea 1. Función de cálculo del pago

```python
def calculo(horas, tarifa):
    if horas <= 40:
        salario = horas * tarifa
    elif horas > 40 and horas <= 50:
        salario = 40 * tarifa + (horas - 40) * tarifa * 1.1
    elif horas > 50:
        salario = 40 * tarifa + 10 * tarifa * 1.1 + (horas - 50) * tarifa * 1.2
    else:
        salario = "Parámetros incorrectos"
    return salario
```

### Tarea 2. Comprobar con el ejemplo

```python
"""
Número de horas trabajadas: 55 



Tarifa por hora: 5                       



Monto a pagar: 40(5) + 10(5)(1.1) + 5(5)(1.2) = 285 
"""
print(calculo(55, 5)) # 285
```

### Tarea 3. Casos de prueba

```python
"""
Caso 1:

Número de horas trabajadas: 35

Tarifa por hora: 10 
"""
print(calculo(35, 10)) 
```

```python
"""
Caso 2:

Número de horas trabajadas: 47

Tarifa por hora: 8 
"""

print(calculo(47, 8)) 
```

```python
"""
Caso 3:

Número de horas trabajadas: 65

Tarifa por hora: 15
"""
print(calculo(65, 15)) 
```
