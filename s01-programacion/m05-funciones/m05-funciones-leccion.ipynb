## 1. Definir y llamar una función

Una función se define con `def`, un nombre, los **parámetros** entre paréntesis y dos puntos. El cuerpo va indentado. Al **llamarla**, los valores que se pasan se llaman **argumentos** y ocupan el lugar de los parámetros.

| Término | Dónde aparece | Ejemplo |
| --- | --- | --- |
| Parámetro | En la definición | `def area(base, altura):` |
| Argumento | En la llamada | `area(4, 2.5)` |
| Valor de retorno | Lo que entrega `return` | `10.0` |

```python
def area_triangulo(base, altura):
    return base * altura / 2

print(area_triangulo(4, 2.5))     # argumentos posicionales
print(area_triangulo(10, 3))
```

## 2. `return` frente a `print`

`print` solo **muestra** un valor en pantalla; `return` lo **entrega** a quien llamó a la función, que puede guardarlo, compararlo o usarlo en otro cálculo. Una función sin `return` devuelve `None`.

```python
def doble_con_print(x):
    print(x * 2)

def doble(x):
    return x * 2

resultado_1 = doble_con_print(5)   # muestra 10, pero no lo devuelve
resultado_2 = doble(5)

print("resultado_1:", resultado_1)
print("resultado_2:", resultado_2)
print("resultado_2 + 1:", resultado_2 + 1)
```

Regla práctica: las funciones de cálculo devuelven con `return` y dejan el `print` a quien las llama.

## 3. Argumentos por defecto y por nombre

Un parámetro puede tener un **valor por defecto**, que se usa cuando la llamada lo omite. Los parámetros con valor por defecto van después de los obligatorios.

Al llamar, los argumentos pueden pasarse **por nombre** (`parametro=valor`). Así el orden deja de importar y la llamada se lee mejor.

```python
def precio_final(precio, descuento=0.0, iva=0.16):
    return round(precio * (1 - descuento) * (1 + iva), 2)

print(precio_final(100))                  # usa los valores por defecto
print(precio_final(100, 0.2))             # descuento posicional
print(precio_final(100, iva=0.08))        # solo cambia el IVA
print(precio_final(iva=0.0, precio=50))   # por nombre, en cualquier orden
```

## 4. Docstrings y anotaciones de tipo

El **docstring** es una cadena entre triples comillas justo debajo del `def`. Explica qué hace la función, qué recibe y qué devuelve, y aparece al consultar `help(función)`.

Las **anotaciones de tipo** (*type hints*) indican el tipo esperado de cada parámetro y del resultado. Python no las hace cumplir: son documentación que también aprovechan los editores.

```python
def promedio(valores: list[float]) -> float:
    """Devuelve la media aritmética de una lista no vacía de números.

    Parámetros:
        valores: lista de números.

    Devuelve:
        La suma de los valores dividida entre su cantidad.
    """
    return sum(valores) / len(valores)

print(round(promedio([7.5, 9.0, 8.0]), 2))
print(promedio.__doc__.splitlines()[0])   # primera línea del docstring
```

## 5. Alcance local y global

Las variables creadas dentro de una función son **locales**: existen solo mientras la función se ejecuta y no afectan a variables de fuera con el mismo nombre. Una función puede **leer** una variable global, pero si le asigna un valor dentro, Python crea una local nueva.

Depender de variables globales hace que el resultado cambie según el estado del programa. Lo recomendable es pasar todo lo que la función necesita como parámetros.

```python
tasa = 0.10          # variable global

def interes(capital):
    ganancia = capital * tasa    # lee la global; ganancia es local
    return ganancia

print(interes(1000))

tasa = 0.25          # cambiar la global cambia el resultado
print(interes(1000))

def interes_explicito(capital, tasa):   # mejor: todo entra como parámetro
    return capital * tasa

print(interes_explicito(1000, 0.10))
```

## 6. Slicing con paso

El slicing `secuencia[inicio:fin:paso]` funciona igual en listas y cadenas. Con un **paso negativo** recorre la secuencia de derecha a izquierda; si se omiten inicio y fin, `[::-1]` devuelve la secuencia completa invertida.

| Expresión | Con `s = "python"` |
| --- | --- |
| `s[::2]` | `"pto"` |
| `s[1::2]` | `"yhn"` |
| `s[::-1]` | `"nohtyp"` |
| `s[-3:]` | `"hon"` |

```python
s = "python"
print(s[::2], s[1::2], s[::-1], s[-3:])

# Funciona igual con listas
print([1, 2, 3, 4, 5][::-1])

# Comprobar si una palabra es palíndromo, sin distinguir mayúsculas
palabra = "Reconocer"
print(palabra.lower() == palabra.lower()[::-1])
```

## 7. Primalidad: probar divisores hasta $\sqrt{n}$

Un número **primo** es un natural mayor que 1 cuyos únicos divisores positivos son 1 y él mismo. Para comprobarlo basta con buscar divisores entre 2 y $\sqrt{n}$: si $n = a \cdot b$ con $a \le b$, entonces $a \le \sqrt{n}$. Si no hay ningún divisor hasta ahí, no lo hay en absoluto.

Esto reduce mucho el trabajo: para $n \approx 10^6$ se prueban unos mil divisores en lugar de un millón. `math.isqrt(n)` da la raíz cuadrada entera exacta, sin errores de redondeo.

Los casos pequeños deben tratarse aparte: 0 y 1 **no** son primos, y 2 es el único primo par.

```python
from math import isqrt

def es_primo(n: int) -> bool:
    """True si n es primo."""
    if n < 2:                   # 0, 1 y negativos
        return False
    for d in range(2, isqrt(n) + 1):
        if n % d == 0:
            return False
    return True                 # con n = 2 o 3 el rango está vacío y se llega aquí

print([k for k in range(30) if es_primo(k)])
print(isqrt(97), es_primo(97))
```

## 8. Probar una función con `assert`

`assert condición` no hace nada si la condición es verdadera y lanza un `AssertionError` si es falsa. Reunir varios `assert` con casos típicos y **casos borde** es una forma rápida de comprobar que una función hace lo que debe y de detectar si un cambio posterior la rompe.

```python
casos = {0: False, 1: False, 2: True, 3: True, 4: False, 9: False, 25: False, 29: True, 7919: True}

for n, esperado in casos.items():
    assert es_primo(n) == esperado, f"Falla con n={n}"
print("Todas las pruebas pasaron:", len(casos), "casos")

# Así se ve un assert que falla (capturado para que el notebook continúe)
try:
    assert es_primo(1) == True, "1 no debería ser primo"
except AssertionError as error:
    print("AssertionError:", error)
```

## 9. Errores frecuentes

**Usar `print` en lugar de `return`.** La función parece funcionar porque el valor aparece en pantalla, pero devuelve `None` y falla en cuanto se usa en otro cálculo (ver sección 2).

**Usar una lista como valor por defecto.** El valor por defecto se crea **una sola vez**, al definir la función. Si es mutable, como una lista, los cambios se acumulan entre llamadas. La forma correcta es usar `None` y crear la lista dentro.

```python
def agregar_mal(elemento, lista=[]):
    lista.append(elemento)
    return lista

print(agregar_mal("a"))
print(agregar_mal("b"))     # conserva "a" de la llamada anterior

def agregar(elemento, lista=None):
    if lista is None:
        lista = []
    lista.append(elemento)
    return lista

print(agregar("a"))
print(agregar("b"))
```

**Olvidar el `+ 1` en el límite de la raíz.** `range(2, isqrt(n))` excluye la propia raíz, así que los cuadrados de primos, como 25 o 49, se clasifican como primos:

```python
def es_primo_mal(n):
    if n < 2:
        return False
    for d in range(2, isqrt(n)):      # no prueba d = isqrt(n)
        if n % d == 0:
            return False
    return True

print([k for k in [4, 9, 25, 49] if es_primo_mal(k)])   # todos son compuestos
print([k for k in [4, 9, 25, 49] if es_primo(k)])
```

**Llamar a una función antes de definirla.** En un notebook, la celda con el `def` debe ejecutarse antes que la celda que la usa; si no, aparece un `NameError`.

## 10. Ejercicio

Un número es **perfecto** si es igual a la suma de sus divisores propios (los divisores positivos menores que él). Por ejemplo, $6 = 1 + 2 + 3$.

1. Escribe `suma_divisores(n: int) -> int` con docstring, que devuelva la suma de los divisores propios de `n`. Recorre solo hasta $\sqrt{n}$: cada divisor $d$ encontrado aporta también su pareja $n / d$ (sin contarla dos veces si $d = n / d$).
2. Escribe `es_perfecto(n: int) -> bool`, que devuelva `False` para $n < 2$.
3. Añade a `es_perfecto` un parámetro `mostrar=False`: si es verdadero, la función imprime la suma además de devolver el resultado.
4. Comprueba con `assert` que 6, 28 y 496 son perfectos y que 0, 1, 2, 12 y 27 no lo son.
5. Lista los números perfectos menores que 10 000.

### Solución

```python
def suma_divisores(n: int) -> int:
    """Suma de los divisores positivos de n menores que n (n >= 2)."""
    total = 1                           # 1 divide a todo n >= 2
    for d in range(2, isqrt(n) + 1):
        if n % d == 0:
            pareja = n // d
            total += d if d == pareja else d + pareja
    return total

def es_perfecto(n: int, mostrar: bool = False) -> bool:
    """True si n es igual a la suma de sus divisores propios."""
    if n < 2:
        return False
    s = suma_divisores(n)
    if mostrar:
        print(f"{n}: suma de divisores propios = {s}")
    return s == n

for n in [6, 28, 496]:
    assert es_perfecto(n)
for n in [0, 1, 2, 12, 27]:
    assert not es_perfecto(n)
print("Pruebas superadas")

es_perfecto(12, mostrar=True)
print([n for n in range(10_000) if es_perfecto(n)])
```

## Resumen

- Una función agrupa lógica con nombre: recibe **argumentos** en sus **parámetros** y entrega un resultado con `return`, que no es lo mismo que mostrarlo con `print`.
- Los argumentos por defecto y por nombre hacen las llamadas más flexibles; los valores por defecto mutables se sustituyen por `None`.
- Docstrings y anotaciones de tipo documentan qué recibe y qué devuelve cada función.
- Las variables internas son locales; conviene pasar los datos como parámetros en lugar de depender de globales.
- Para probar la primalidad basta con buscar divisores hasta $\sqrt{n}$, tratando aparte 0, 1 y 2; una batería de `assert` con casos borde confirma que la función es correcta.
