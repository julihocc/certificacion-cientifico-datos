## Contexto

Contar la frecuencia de las palabras es el primer paso de muchos análisis de texto. Aquí se cuenta, con un diccionario y sin librerías, cuántas veces aparece cada palabra de un trabalenguas.

### Tarea 1. Contar las palabras de una frase

```python
texto = "El amor es una locura que ni el cura lo cura porque si el cura lo cura sería una locura del cura"

conteo = {}

tokens = texto.split()

for token in tokens:
    if token in conteo:
        conteo[token] += 1
    else:
        conteo[token] = 1

for palabra in conteo:
    print(
        f"La palabra ' {palabra} ' re repite {conteo[palabra]} instancia(s)."
    )
```
