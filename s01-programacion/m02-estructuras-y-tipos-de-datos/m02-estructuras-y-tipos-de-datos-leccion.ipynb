## 1. Listas

Una **lista** es una secuencia ordenada y **mutable**: se puede modificar después de crearla. Se escribe entre corchetes y puede mezclar tipos, aunque lo habitual es que todos sus elementos sean del mismo tipo.

Cada elemento tiene una posición (índice) que empieza en 0. Los índices negativos cuentan desde el final, y el **slicing** `lista[inicio:fin:paso]` extrae una sublista sin incluir la posición `fin`.

````python
temperaturas = [21.5, 23.0, 19.8, 25.1, 22.4]

print(temperaturas[0])     # primer elemento
print(temperaturas[-1])    # último elemento
print(temperaturas[1:3])   # posiciones 1 y 2 (la 3 no se incluye)
print(temperaturas[::-1])  # toda la lista al revés
print(len(temperaturas))   # número de elementos
````

Los métodos de lista modifican la lista **en el lugar** y devuelven `None`:

| Método | Efecto |
| --- | --- |
| `append(x)` | Agrega `x` al final |
| `insert(i, x)` | Inserta `x` en la posición `i` |
| `remove(x)` | Elimina la primera aparición de `x` |
| `pop(i)` | Elimina y devuelve el elemento en `i` (el último si se omite) |
| `sort()` | Ordena la lista |

````python
temperaturas.append(20.0)
temperaturas.sort()
print(temperaturas)

ultima = temperaturas.pop()
print("Se retiró:", ultima)
print(f"Promedio: {sum(temperaturas) / len(temperaturas):.2f}")
````

## 2. Tuplas y desempaquetado

Una **tupla** es como una lista pero **inmutable**: una vez creada no cambia. Se usa para agrupar valores que van juntos, como las coordenadas de un punto o un registro (nombre, edad).

El **desempaquetado** asigna cada elemento de una tupla a una variable en una sola línea.

````python
punto = (19.43, -99.13)          # latitud y longitud
latitud, longitud = punto        # desempaquetado
print(f"Latitud: {latitud}, longitud: {longitud}")

# Una función puede devolver varios valores en una tupla
def minimo_y_maximo(valores):
    return min(valores), max(valores)

bajo, alto = minimo_y_maximo([7, 3, 9, 1])
print(bajo, alto)
````

## 3. Diccionarios

Un **diccionario** asocia **claves** con **valores**. Buscar por clave es inmediato, sin importar cuántos elementos tenga. Las claves deben ser inmutables (texto, números, tuplas) y únicas.

| Operación | Resultado |
| --- | --- |
| `d[clave]` | Valor de la clave; `KeyError` si no existe |
| `d.get(clave, defecto)` | Valor de la clave, o `defecto` si no existe |
| `clave in d` | `True` si la clave existe |
| `d.items()` | Pares `(clave, valor)` para recorrerlos |

````python
capitales = {"México": "Ciudad de México", "Perú": "Lima", "Chile": "Santiago"}

print(capitales["Perú"])
print(capitales.get("Brasil", "desconocida"))
print("Chile" in capitales)

capitales["Brasil"] = "Brasilia"   # agregar o actualizar una clave
for pais, capital in capitales.items():
    print(f"{pais}: {capital}")
````

## 4. Conjuntos

Un **conjunto** (`set`) guarda elementos **sin repetir y sin orden**. Sirve para eliminar duplicados y para operaciones de teoría de conjuntos: unión `|`, intersección `&` y diferencia `-`.

````python
curso_python = {"Ana", "Luis", "Marta", "Pedro"}
curso_sql = {"Luis", "Pedro", "Sofía"}

# sorted() solo para mostrar el resultado en un orden estable
print("En ambos cursos:", sorted(curso_python & curso_sql))
print("En alguno:", sorted(curso_python | curso_sql))
print("Solo en Python:", sorted(curso_python - curso_sql))

# Eliminar duplicados de una lista
colores = ["rojo", "azul", "rojo", "verde", "azul"]
print(sorted(set(colores)))
````

## 5. Cadenas de texto como secuencias

Una cadena se comporta como una secuencia de caracteres: admite índices, slicing y `len`, pero es inmutable. Dos métodos son clave para procesar texto:

- `lower()` devuelve el texto en minúsculas, para que "El" y "el" se consideren iguales.
- `split()` divide el texto en una lista de palabras usando los espacios como separador.

````python
frase = "El dato es el nuevo petróleo, dice el refrán"

print(frase[:7])
print(frase.lower().split())

# Quitar signos de puntuación antes de separar
limpia = frase.lower().replace(",", "")
palabras = limpia.split()
print(palabras)
````

## 6. El patrón de conteo

Contar cuántas veces aparece cada elemento es una de las tareas más comunes. La idea es recorrer los elementos y llevar un diccionario cuyo valor es el conteo de cada clave.

Hay tres formas, de la más explícita a la más compacta:

````python
# a) Con if/else
conteo = {}
for palabra in palabras:
    if palabra in conteo:
        conteo[palabra] += 1
    else:
        conteo[palabra] = 1
print(conteo)

# b) Con dict.get: si la clave no existe, parte de 0
conteo = {}
for palabra in palabras:
    conteo[palabra] = conteo.get(palabra, 0) + 1
print(conteo)
````

````python
# c) Con collections.Counter, de la librería estándar
from collections import Counter

conteo = Counter(palabras)
print(conteo)
print(conteo.most_common(2))   # las dos palabras más frecuentes
````

## 7. Errores frecuentes

**Copiar una lista con `=`.** La asignación no copia la lista: crea otro nombre para el mismo objeto, así que modificar uno modifica el otro. Para una copia independiente se usa `lista.copy()` o `list(lista)`.

````python
original = [1, 2, 3]
alias = original          # mismo objeto
copia = original.copy()   # objeto nuevo

alias.append(4)
print("original:", original)
print("copia:   ", copia)
````

**Acceder a una clave que no existe.** `d[clave]` lanza `KeyError`. Cuando la clave puede faltar, conviene usar `d.get(clave, defecto)` o comprobar antes con `in`.

**Contar sin normalizar el texto.** Sin `lower()` ni limpieza de puntuación, "Dato", "dato" y "dato," se cuentan como tres palabras distintas:

````python
texto = "Dato tras dato, el dato manda"
print(Counter(texto.split()))
print(Counter(texto.lower().replace(",", "").split()))
````

**Esperar orden en un conjunto.** Un `set` no garantiza ningún orden. Si el orden importa, se convierte con `sorted(conjunto)`.

## 8. Ejercicio

Una tienda registra los productos vendidos en el día:

```python
ventas = ["café", "pan", "Café", "leche", "pan", "café", "jugo", "Pan", "café"]
```

1. Normaliza los nombres a minúsculas.
2. Obtén la lista de productos distintos, en orden alfabético.
3. Cuenta cuántas unidades se vendieron de cada producto.
4. Muestra el producto más vendido y sus unidades usando desempaquetado.

### Solución

````python
ventas = ["café", "pan", "Café", "leche", "pan", "café", "jugo", "Pan", "café"]

normalizadas = []
for producto in ventas:
    normalizadas.append(producto.lower())
print("Productos distintos:", sorted(set(normalizadas)))

unidades = {}
for producto in normalizadas:
    unidades[producto] = unidades.get(producto, 0) + 1
print("Unidades:", unidades)

producto, cantidad = Counter(normalizadas).most_common(1)[0]
print(f"Más vendido: {producto} ({cantidad} unidades)")
````

## Resumen

- Las **listas** son secuencias ordenadas y mutables; las **tuplas**, secuencias inmutables que se desempaquetan con facilidad.
- Los **diccionarios** asocian claves con valores; `get` evita errores cuando una clave falta.
- Los **conjuntos** eliminan duplicados y permiten unión, intersección y diferencia.
- Para contar elementos se usa un diccionario con `get` o, más directo, `collections.Counter`.
- Antes de contar texto hay que normalizarlo: minúsculas y sin puntuación.
