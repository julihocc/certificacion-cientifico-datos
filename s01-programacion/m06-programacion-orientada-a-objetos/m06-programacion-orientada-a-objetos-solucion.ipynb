## Contexto

Para procesar números romanos hay que convertirlos a enteros: cada símbolo tiene un valor (`I` = 1, `V` = 5, `X` = 10, `L` = 50, `C` = 100, `D` = 500, `M` = 1000) que se suma, salvo cuando va seguido de uno mayor, en cuyo caso se resta. Esa lógica se agrupa en una clase.

## Instrucciones

1. **La clase `Conversion`.** Escribir la clase, con la tabla de equivalencias como atributo y un método `romano_entero` que convierta un número romano a entero.
2. **Probar la clase** con un objeto llamado `numero`: `numero.romano_entero('MMMCMLXXXVI')` debe devolver 3986.

### Tarea 1. La clase `Conversion`

```python
class Conversion:
    def __init__(self):
        self.romanos = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
    
    def romano_entero(self, romano):
        entero = 0
        i = 0
        while i < len(romano):
            # Si el símbolo actual tiene un valor menor que su siguiente símbolo, restamos el valor actual.
            if i + 1 < len(romano) and self.romanos[romano[i]] < self.romanos[romano[i+1]]:
                entero += self.romanos[romano[i+1]] - self.romanos[romano[i]]
                i += 2
            else:
                entero += self.romanos[romano[i]]
                i += 1
        return entero
```

### Tarea 2. Probar la clase

```python

# Ejemplo de uso
numero = Conversion()
print(numero.romano_entero('MMMCMLXXXVI'))
```
