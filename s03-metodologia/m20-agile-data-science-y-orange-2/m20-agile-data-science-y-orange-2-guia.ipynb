# Módulo 20: Agile Data Science y Orange (2) — Guía del ejercicio

En este ejercicio vas a comparar la gestión tradicional y la gestión ágil de un proyecto de ciencia de datos y, después, a construir en **Orange** un modelo de clasificación que recomiende un medicamento, comparando modelos y calibraciones hasta encontrar el que mejor predice.

- Cuando termines, compara tu trabajo con la solución: el reporte [m20-agile-data-science-y-orange-2-solucion.md](m20-agile-data-science-y-orange-2-solucion.md) y el flujo de Orange [modelo.ows](modelo.ows).

## Contexto

En ciencia de datos es difícil saber de antemano qué datos servirán o qué modelo funcionará; por eso muchos equipos trabajan de forma ágil, en ciclos cortos con revisiones frecuentes, en lugar de planificarlo todo al inicio. La primera parte del ejercicio te pide reflexionar sobre esa diferencia con un caso real.

En la segunda parte eres un investigador médico. Recopilaste datos de pacientes con la misma enfermedad; durante el tratamiento, cada uno respondió a uno de cinco medicamentos: los fármacos A, B y C (de proveedor nacional) y los fármacos X e Y (de proveedor extranjero). Quieres un modelo que indique qué medicamento podría ser apropiado para un paciente futuro con la misma enfermedad.

## Datos

Los archivos están en esta misma carpeta:

- `drugs.csv`: 200 pacientes, separado por comas, con una fila por paciente.
- `drugs-prueba.csv`: los dos pacientes nuevos que debes clasificar (sin la columna `Drug`). Empieza con una marca BOM de UTF-8; Orange la lee sin problema.

| Columna | Descripción |
| --- | --- |
| `Age` | Edad (años) |
| `Sex` | Sexo: `F` o `M` |
| `BP` | Presión arterial: `LOW`, `NORMAL` o `HIGH` |
| `Cholesterol` | Colesterol: `NORMAL` o `HIGH` |
| `Na_to_K` | Razón sodio/potasio en sangre |
| `Drug` | Variable objetivo: `drugA`, `drugB`, `drugC`, `drugX` o `drugY` |

Los pacientes nuevos son:

| Age | Sex | BP | Cholesterol | Na_to_K |
| --- | --- | --- | --- | --- |
| 40 | M | LOW | HIGH | 24.222 |
| 55 | F | HIGH | NORMAL | 11.150 |
