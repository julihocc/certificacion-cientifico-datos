## Contexto

Las preguntas de negocio rara vez se responden con una consulta sobre una sola tabla. En AdventureWorks, saber qué productos se han vendido exige mirar las líneas de venta; saber en qué departamento trabaja hoy cada empleado exige recorrer su historial de departamentos y quedarse con el registro vigente; y preparar un listado de órdenes para otro sistema exige filtrar por año y recortar códigos.

Vas a resolver esas tareas combinando subconsultas, CTE, tablas temporales y funciones integradas de SQL Server.
