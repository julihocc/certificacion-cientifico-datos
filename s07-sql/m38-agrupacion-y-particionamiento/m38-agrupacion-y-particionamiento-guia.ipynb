## Contexto

El equipo comercial de AdventureWorks quiere saber qué productos se venden más y cuánto pesa cada línea dentro de su pedido. Para lo primero hay que **agrupar** las líneas de venta por producto y sumar cantidades e importes; para lo segundo hay que **particionar** por pedido y calcular, junto a cada línea, el total del pedido o la posición de la línea dentro de él.

Todas las preguntas se responden con una sola tabla, `Sales.SalesOrderDetail`, que tiene una fila por línea de pedido.

## Datos

Trabajarás con la tabla `Sales.SalesOrderDetail` de la base `AdventureWorks2022`. Las columnas que necesitas son:
