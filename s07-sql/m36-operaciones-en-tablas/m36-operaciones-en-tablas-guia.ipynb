## Contexto

Una tienda en línea quiere registrar sus ventas en una tabla propia dentro de su servidor SQL Server. Necesita guardar cada orden con sus fechas, su estatus, los datos de entrega, el método de pago y de envío, y los importes. Con el tiempo, la tabla tendrá que cambiar: se renombrará, ganará columnas y, cuando la empresa se expanda a otros estados, el número de orden deberá llevar un prefijo con la clave del estado de la sucursal donde se originó.

Vas a construir esa tabla desde cero y aplicarle esos cambios con código SQL, comprobando el resultado después de cada paso.
