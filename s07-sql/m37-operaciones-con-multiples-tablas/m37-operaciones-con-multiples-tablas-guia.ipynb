## Contexto

En una base de datos relacional la información de una misma entidad se reparte entre varias tablas. En AdventureWorks, por ejemplo, los datos laborales de un empleado están en `HumanResources.Employee`, pero su nombre está en `Person.Person`; las órdenes de venta guardan solo el identificador del cliente, y las reseñas de productos viven en una tabla aparte. Para responder preguntas sencillas como «¿cómo se llama cada empleado y cuál es su puesto?» hay que unir tablas.

El tipo de unión decide qué filas sobreviven cuando no hay coincidencia, y elegir mal cambia el resultado sin dar ningún error. Por eso cada tarea te pide un tipo de unión concreto: fíjate en qué filas debe conservar.
