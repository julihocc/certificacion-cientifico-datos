# Módulo 32: Estadística descriptiva — Guía del ejercicio

En este ejercicio vas a practicar las herramientas básicas de la estadística descriptiva: leer un diagrama de caja, calcular medidas de tendencia central y de dispersión, reconocer tipos de correlación en gráficos de dispersión e interpretar un modelo de regresión lineal simple.

- Cuando termines, compara tu trabajo con la solución: [m32-estadistica-descriptiva-solucion.md](m32-estadistica-descriptiva-solucion.md).

## Contexto

Antes de aplicar cualquier modelo, un científico de datos resume y describe los datos: dónde se concentran, cuánto varían, si hay valores atípicos y cómo se relacionan dos variables entre sí. Estos cálculos e interpretaciones aparecen en cualquier informe, desde un tablero de ventas hasta un estudio clínico.

Este ejercicio reúne cuatro situaciones breves, como las que encontrarías al revisar el informe de un colega: un diagrama de caja de las estaturas de los habitantes de una ciudad, un pequeño conjunto de datos para resumir, cinco gráficos de dispersión y una recta de regresión con su coeficiente de determinación.
