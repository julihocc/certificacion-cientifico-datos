## 3. Variables y tipos de datos básicos

Una variable es un nombre que apunta a un valor. Python infiere el tipo a partir del valor, y la función `type()` nos dice cuál es.

````python
edad = 34                 # int: número entero
estatura = 1.75           # float: número con decimales
nombre = "Ana"            # str: texto
es_estudiante = True      # bool: verdadero o falso
telefono = None           # NoneType: ausencia de valor

for valor in [edad, estatura, nombre, es_estudiante, telefono]:
    print(repr(valor), "->", type(valor).__name__)
````

Los nombres de variables distinguen mayúsculas de minúsculas, no pueden empezar con un número y, por convención, se escriben en `snake_case`. Una variable puede reasignarse a un valor de otro tipo:

````python
x = 10
print(type(x).__name__)
x = "diez"
print(type(x).__name__)
````

## 4. Operadores

### Aritméticos

````python
a, b = 17, 5

print("suma:          ", a + b)
print("resta:         ", a - b)
print("multiplicación:", a * b)
print("división:      ", a / b)   # siempre devuelve float
print("división entera:", a // b)
print("residuo:       ", a % b)
print("potencia:      ", a ** 2)
````

### De comparación y lógicos

Las comparaciones devuelven un `bool`, y los operadores `and`, `or` y `not` combinan condiciones.

````python
temperatura = 38.2
tiene_tos = True

fiebre = temperatura >= 38
print("¿Fiebre?", fiebre)
print("¿Fiebre y tos?", fiebre and tiene_tos)
print("¿Sin fiebre?", not fiebre)
print("¿Temperatura normal?", 36 <= temperatura < 37.5)  # comparaciones encadenadas
````

## 5. Cadenas de texto

Las cadenas (`str`) se pueden indexar, recortar y transformar con sus métodos. Las *f-strings* permiten insertar valores dentro del texto.

````python
frase = "Ciencia de Datos"

print(frase[0], frase[-1])     # primer y último carácter
print(frase[:7])               # recorte: los primeros 7 caracteres
print(frase.upper())
print(frase.replace("Datos", "la Computación"))
print(len(frase), "caracteres")

precio = 1234.5678
print(f"{nombre} tiene {edad} años y el precio es ${precio:,.2f}")
````

## 6. Conversión entre tipos

Los datos suelen llegar como texto (por ejemplo, al leer un archivo), así que es común convertirlos con `int()`, `float()`, `str()` y `bool()`.

````python
texto = "42"
numero = int(texto)
print(numero + 8)

print(float("3.1416") * 2)
print(str(2024) + " fue bisiesto")
print(bool(0), bool(1), bool(""), bool("hola"))  # valores 'vacíos' son False
````

````python
# Convertir texto que no representa un número produce un error
try:
    int("cuarenta y dos")
except ValueError as error:
    print("ValueError:", error)
````

## 7. Ejercicio

El índice de masa corporal se calcula como $IMC = \dfrac{peso}{estatura^2}$, con el peso en kg y la estatura en metros. Los datos llegan como texto:

```python
peso_texto = "72.5"
estatura_texto = "1.68"
```

1. Convierte ambos valores a números y calcula el IMC.
2. Imprime el resultado con dos decimales usando una f-string.
3. Imprime `True` si el IMC está en el rango saludable, de 18.5 a menos de 25.

### Solución

````python
peso_texto = "72.5"
estatura_texto = "1.68"

peso = float(peso_texto)
estatura = float(estatura_texto)
imc = peso / estatura ** 2

print(f"IMC: {imc:.2f}")
print("¿Rango saludable?", 18.5 <= imc < 25)
````
