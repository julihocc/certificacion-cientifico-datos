## Contexto

Una cadena de tiendas de renta de DVD quiere conocer mejor su catálogo, a sus clientes y sus ventas: qué actores participan en más películas, cuánto cuesta rentar cada título, qué películas se rentan más y cuáles nunca, cuántos clientes tiene cada tienda y cuánto vende cada una. Toda esa información está en su base de datos relacional, y tu trabajo es responder cada pregunta con una consulta de SQL.

## Datos

Se usa **dvdrental**, la base de datos de ejemplo de PostgreSQL. Las tablas que necesitas son:

| Tabla | Contenido | Columnas relevantes |
| --- | --- | --- |
| `actor` | Actores | `actor_id`, `first_name`, `last_name` |
| `film` | Películas | `film_id`, `title`, `rental_rate`, `rating` (`G`, `PG`, `PG-13`, `R`, `NC-17`) |
| `film_actor` | Relación entre actores y películas | `actor_id`, `film_id` |
| `inventory` | Copias físicas de cada película en cada tienda | `inventory_id`, `film_id`, `store_id` |
| `rental` | Rentas | `rental_id`, `rental_date`, `inventory_id`, `customer_id` |
| `payment` | Pagos de las rentas | `payment_id`, `rental_id`, `customer_id`, `amount` |
| `customer` | Clientes | `customer_id`, `store_id`, `first_name`, `last_name` |
| `store` | Tiendas | `store_id` |

Observa que una renta no apunta a una película, sino a una copia del inventario: para llegar de `rental` a `film` tienes que pasar por `inventory`.

**Cómo restaurar la base.** La carpeta `dvdrental/` contiene `restore.sql`, un volcado de `pg_dump` en texto plano que crea el esquema y carga los datos desde archivos `.dat` que **no** están en el repositorio. Descarga el archivo `dvdrental.tar` de la [base de datos de ejemplo de PostgreSQL Tutorial](https://www.postgresqltutorial.com/postgresql-getting-started/postgresql-sample-database/) y restáuralo con `pg_restore -U postgres -d dvdrental dvdrental.tar` después de crear la base vacía (`CREATE DATABASE dvdrental;`). Si prefieres usar `restore.sql`, extrae los `.dat` del `.tar` y sustituye en el script la ruta `/home/julihocc/...` por la de tu carpeta; ten en cuenta que el script empieza con `DROP DATABASE dvdrental;`.
