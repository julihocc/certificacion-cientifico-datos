## Datos

Los dos archivos están en esta misma carpeta.

**`apps.csv`**: una fila por aplicación (9 659 filas) con 13 características.

| Columna | Descripción |
| --- | --- |
| `App` | Nombre de la aplicación |
| `Category` | Categoría (por ejemplo, `FAMILY`, `GAME`, `TOOLS`) |
| `Rating` | Calificación promedio de 1 a 5 (con faltantes) |
| `Reviews` | Número de reseñas |
| `Size` | Tamaño en MB (con faltantes) |
| `Installs` | Instalaciones, en texto con comas y `+` (por ejemplo, `"10,000+"`) |
| `Type` | `Free` o `Paid` |
| `Price` | Precio, en texto con `$` (por ejemplo, `"$4.99"`; `"0"` si es gratuita) |
| `Content Rating`, `Genres` | Público objetivo y géneros |
| `Last Updated`, `Current Ver`, `Android Ver` | Fecha de la última actualización, versión actual y versión mínima de Android |

**`user-reviews.csv`**: hasta 100 reseñas por aplicación (64 295 filas). Cada reseña se procesó previamente y tiene tres características: sentimiento, polaridad y subjetividad.

| Columna | Descripción |
| --- | --- |
| `App` | Nombre de la aplicación (sirve para unir con `apps.csv`) |
| `Review` | Texto de la reseña (con faltantes) |
| `Sentiment` | `Positive`, `Negative` o `Neutral` (con faltantes) |
| `Sentiment_Polarity` | Polaridad del sentimiento, de −1 (negativo) a 1 (positivo) |
| `Sentiment_Subjectivity` | Subjetividad, de 0 (objetivo) a 1 (subjetivo) |
