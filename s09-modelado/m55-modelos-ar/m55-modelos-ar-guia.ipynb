## Contexto

Un analista de una casa de bolsa sigue la acción de **The Walt Disney Company** (`DIS`) y quiere un modelo sencillo para anticipar su precio de cierre a corto plazo. Con los datos del primer trimestre de 2023 te pide ajustar un modelo autorregresivo, medir qué tan bien habría pronosticado las últimas semanas del trimestre y proyectar el precio para las semanas siguientes, a partir de abril, con un rango de confianza del 90 %.

Un modelo AR($p$) explica el valor actual de la serie con sus $p$ valores anteriores:

$$y_t = c + \phi_1 y_{t-1} + \phi_2 y_{t-2} + \dots + \phi_p y_{t-p} + \varepsilon_t$$
